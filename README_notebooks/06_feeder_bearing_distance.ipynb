## Identify bearing and distances of feeder from hive

```python
import math

def dms_to_decimal(degrees, minutes, seconds, hemisphere):
    decimal = degrees + minutes / 60 + seconds / 3600
    if hemisphere in ["S", "W"]:
        decimal *= -1
    return decimal

def initial_bearing(lat1, lon1, lat2, lon2):
    """
    Calculates initial bearing from point 1 to point 2.
    Bearing is returned in degrees clockwise from geographic north.
    """
    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)
    delta_lambda = math.radians(lon2 - lon1)

    x = math.sin(delta_lambda) * math.cos(phi2)
    y = (
        math.cos(phi1) * math.sin(phi2)
        - math.sin(phi1) * math.cos(phi2) * math.cos(delta_lambda)
    )

    bearing = math.degrees(math.atan2(x, y))
    return (bearing + 360) % 360

def haversine_distance(lat1, lon1, lat2, lon2):
    """
    Calculates distance between two coordinates in metres.
    """
    R = 6371000  # Earth radius in metres

    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)
    delta_phi = math.radians(lat2 - lat1)
    delta_lambda = math.radians(lon2 - lon1)

    a = (
        math.sin(delta_phi / 2) ** 2
        + math.cos(phi1) * math.cos(phi2) * math.sin(delta_lambda / 2) ** 2
    )

    return 2 * R * math.atan2(math.sqrt(a), math.sqrt(1 - a))


# Replace these with your exact coordinates
hive_lat = dms_to_decimal(25, 34, 18, "S")
hive_lon = dms_to_decimal(31, 10, 48.2, "E")

feeder_lat = dms_to_decimal(25, 34, 13.6, "S")
feeder_lon = dms_to_decimal(31, 10, 57.1, "E")

bearing = initial_bearing(hive_lat, hive_lon, feeder_lat, feeder_lon)
distance = haversine_distance(hive_lat, hive_lon, feeder_lat, feeder_lon)

print(f"Bearing: {bearing:.1f}°")
print(f"Distance: {distance:.1f} m")
```
